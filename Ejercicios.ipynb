# Taller de Análisis de Datos

```python
import pandas as pd
```

## 1. Carga y verificación inicial

```python
#cargar el csv en un dataframe (sin columnas indice adicionales)
df_ventas = pd.read_csv("cafeterias_especialidad - cafeterias_especialidad.csv", index_col=False) #index_col=False para que no use ninguna columna como indice

df_ventas.head()
```

```python
#verificacion inicial: tamaño (filas y columnas)
df_ventas.shape
```

```python
#verificacion inicial: cantidad de nulos por columna
df_ventas.isna().sum()
```

## 2. Selección de variables clave

```python
#seleccionar una sola columna como Series
metodos = df_ventas["Metodo_Extraccion"]
metodos, type(metodos) #muestra la serie y confirma que es un pandas Series
```

```python
#seleccionar varias columnas (dataframe)
df_ventas[["Sede", "Variedad", "Precio_Taza"]]
```

## 3. Localización de una transacción específica

```python
#filtrar por indice con loc
df_ventas.loc[45] #muestra toda la informacion del registro en el indice 45
```

## 4. Depuración de registros duplicados

```python
#identificar cuantos registros estan duplicados
df_ventas.duplicated().sum() #cantidad de filas duplicadas
```

```python
#eliminar duplicados
df_ventas = df_ventas.drop_duplicates() #elimina las filas duplicadas
df_ventas.shape
```

## 5. Filtrado por umbral de satisfacción

```python
#filtrar por condicion
df_ventas[df_ventas["Calificacion_Cliente"] >= 4.5] #ventas con calificacion mayor o igual a 4.5
```

## 6. Filtrado por lista de opciones (isin)

```python
#filtrar por lista de valores (isin)
df_ventas[df_ventas["Sede"].isin(["Manizales", "Pereira"])] #solo pedidos de Manizales y Pereira
```

## 7. Filtrado numérico por intervalo (between)

```python
#filtrar por intervalo
df_ventas[df_ventas["Precio_Taza"].between(7500, 9500)] #between es inclusivo por defecto
```

## 8. Consulta estructurada con .query()

```python
#filtrar por query
df_ventas.query("Sede == 'Bogotá' and Cantidad >= 3")
```

## 9. Filtrado compuesto con conectores lógicos

```python
#filtrar por condicion y conectores logicos
filtro = (df_ventas["Variedad"] == "Geisha") & (df_ventas["Metodo_Extraccion"].str.startswith("C")) #variedad Geisha y metodo que empieza con C
df_ventas[filtro]
```

## 10. Tratamiento de valores faltantes

```python
#rellenar valores nulos con un valor especifico
df_ventas = df_ventas.fillna(
    {
        "Precio_Taza": df_ventas["Precio_Taza"].mean(), #rellena los nulos con la media de la columna Precio_Taza
        "Calificacion_Cliente": 4.0 #rellena los nulos con el valor constante 4.0
    }
)

df_ventas.isna().sum() #verificamos que ya no hay nulos en esas columnas
```

## 11. Transformación con funciones y creación de columnas (apply)

```python
#transformar datos
def calcular_subtotal(fila):
    return fila["Precio_Taza"] * fila["Cantidad"]

df_ventas["Total_Venta"] = df_ventas.apply(calcular_subtotal, axis=1) #axis=1 para aplicar la funcion fila por fila
df_ventas.head()
```

## 12. Estandarización y reorganización de columnas

```python
#renombrar columnas
df_ventas = df_ventas.rename(columns={
    "Metodo_Extraccion": "Metodo",
    "Tipo_Grano": "Tueste"
})
df_ventas.head()
```

```python
#ordenar columnas (Total_Venta en la primera posicion)
columnas_ordenadas = ["Total_Venta"] + [col for col in df_ventas.columns if col != "Total_Venta"] #Total_Venta primero y luego las demas
df_ventas = df_ventas[columnas_ordenadas]
df_ventas.head()
```

## 13. Agregaciones básicas por grupo

```python
#agrupar datos por sede
grouped = df_ventas.groupby("Sede")

#calcular el promedio de la calificacion y la suma de Total_Venta
aggregate_data = grouped.agg({
    "Calificacion_Cliente": "mean",
    "Total_Venta": "sum"
}).round(1) #redondea los valores a 1 decimal
aggregate_data
```

## 14. Métricas personalizadas en agrupamientos

```python
#definir una funcion de agregacion personalizada
def rango(series):
    return series.max() - series.min()

#aplicamos la funcion con agg al grupo
aggregate_data_custom = grouped.agg({
    "Precio_Taza": rango,
    "Total_Venta": rango
})
aggregate_data_custom
```

## 15. Agrupamiento multinivel

```python
#agrupar por sede y tueste
grouped_multi = df_ventas.groupby(["Sede", "Tueste"])

#calcular el promedio de la calificacion y la suma del total de ventas por sede y tueste
aggregate_data_multi = grouped_multi.agg({
    "Calificacion_Cliente": "mean",
    "Total_Venta": "sum"
}).round(1)
aggregate_data_multi
```
